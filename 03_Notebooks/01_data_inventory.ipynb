### Data Inventory

With all tables loaded, we now want to create a data inventory that documents the actual structure of each table

```python

datasets = {
    "customers": customers,
    "date": date,
    "employees": employees,
    "products": products,
    "stores": stores,
    "order_details": order_details,
    "orders_2022_2023": orders_2022_2023,
    "orders_2024": orders_2024,
    "returns": returns
}

for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(f"Table: {name}")
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")
    print("Column names:")
    print(df.columns.tolist())
```

#### Check 2: Data Types and Sample Records

We want to inspect the structure and sample records for each table

```python

for name, df in datasets.items():
    print(f"\n{'=' * 60}")
    print(f"TABLE: {name}")
    print("\nData types:")
    print(df.dtypes)

    print("\nSample records:")
    display(df.head(3))
```

#### Check 3: Check for Missing Values

We now move from inventory to preliminary quality assessment. Here we want to check missing values recognized by pandas.

We are not necessarily identifying empty strings, white-space only text, or placeholders such as `unknown` or `N/A`. We'll investigate these separately.

We're not yet removing records with missing values. A missing email address, for example, may not have bearing on whether we can use that customer's purchase history.


```python

for name, df in datasets.items():
    missing = df.isna().sum()
    missing = missing[missing > 0]

    print(f"\nTable: {name}")

    if missing.empty:
        print("No pandas-recognized missing values.")
    else:
        missing_report = pd.DataFrame({
            "missing_count": missing,
            "missing_pct": (
                missing / len(df) * 100
            ).round(2)
        }).sort_values("missing_pct", ascending=False)

        display(missing_report)
```

#### Check 4: Duplicate Records

We want to check for rows that are identical across all columns. We are not, however, establishing whether the primary keys are duplicated. For instance, two records may share the same `customer_id` or `order_id` while differing in other columns. This will require a separate check.

```python

for name, df in datasets.items():
    print(
        f"{name}: "
        f"{df.duplicated().sum():,} "
        "fully duplicated rows"
    )
```

### Table Grain and Key Integrity

Before we join anything we want to document what one row represents in each table.

We'll use the following as our initial expectations:

|Table	|Expected grain	|Candidate primary key|
|---|---|---|
|Customers	|One row per customer	|`customer_id`|
|Date	|One row per calendar date	|`date_id`|
|Employees	|One row per employee	|`employee_id`|
|Products	|One row per product	|`product_id`|
|Stores	|One row per store	|`store_id`|
|Order details	|One row per order-product line	|`detail_id`|
|Orders, 2022–2023	|One row per order	|`order_id`|
|Orders, 2024	|One row per order	|`order_id`|
|Returns	|One row per return record	|`return_id`|

Now we want to establish whether each table respects its expected grain.

This is arguably one of the most important investigations in our entire data preparation process.

We'll start with the customers table because we already have evidence of duplication.

```python
# Examine customer ID uniqueness

print("Total customer records:", len(customers))

print(
    "Distinct customer IDs:",
    customers["customer_id"].nunique()
)

print(
    "Unique customer IDs:",
    customers["customer_id"].is_unique
)

print(
    "Missing customer IDs:",
    customers["customer_id"].isna().sum()
)

print(
    "Duplicate customer ID records:",
    customers["customer_id"].duplicated().sum()
)
```

We need to understand whether the 50 fully duplicated rows account for all the duplicated customer IDs or whether additional identity problems exist.

Without removing anything just yet, we are interested in inspect the records and distinguish between:
- Completely identical records.
- Records with identical IDs but conflicting attributes.
- Records with missing values that may explain apparent differences.

```python
# Identify all customer IDs appearing more than once

duplicate_customer_ids = customers[
    customers.duplicated(
        subset=["customer_id"],
        keep=False
    )
].sort_values("customer_id")

display(duplicate_customer_ids)
```

#### Check 5: Testing Candidate Primary Keys

We want to ensure our primary keys are non-null and unique within the respective tables

```python

primary_keys = {
    "customers": "customer_id",
    "date": "date_id",
    "employees": "employee_id",
    "products": "product_id",
    "stores": "store_id",
    "order_details": "detail_id",
    "orders_2022_2023": "order_id",
    "orders_2024": "order_id",
    "returns": "return_id"
}

for name, key in primary_keys.items():
    df = datasets[name]

    print(f"\n{name} — {key}")
    print(f"Null key values: {df[key].isna().sum():,}")
    print(f"Distinct key values: {df[key].nunique():,}")
    print(f"Duplicate key rows: {df[key].duplicated().sum():,}")
```

Importantly, we want to also check `order_id` across the two separate order tables. We want to establish that although each table could be having unique IDs internally, no IDs are shared with the other table

```python
combined_order_ids = pd.concat(
    [
        orders_2022_2023["order_id"],
        orders_2024["order_id"]
    ],
    ignore_index=True
)

print("Combined order records:", len(combined_order_ids))
print("Distinct order IDs:", combined_order_ids.nunique())
print("Null order IDs:", combined_order_ids.isna().sum())
print(
    "Repeated IDs across the combined tables:",
    combined_order_ids.duplicated().sum()
)
```

Since no IDs repeat, we have no business investigating further whether there is an actual collision or intentional numbering convention.

#### Check 6: Identifying Potential Foreign Key Problems

We want to check whether transaction records reference customers, products, stores and employees that actually exist.

```python
# We start by checking for customer references
# Combine order tables for inspection only
orders = pd.concat(
    [orders_2022_2023, orders_2024],
    ignore_index=True
)

known_customer_ids = set(
    customers["customer_id"].dropna()
)

referenced_customer_ids = set(
    orders["customer_id"].dropna()
)

unmatched_customer_ids = (
    referenced_customer_ids - known_customer_ids
)

print(
    "Distinct customer IDs referenced by orders "
    "but absent from the customer table:",
    len(unmatched_customer_ids)
)

print("Examples:", list(unmatched_customer_ids)[:10])
```

This confirms that transaction records reference customers that actually exist.

#### Verifying the Observation Period

Our planned classification and regression models require reliable historical and future periods.

So we'll start by inspecting the transaction date ranges.

```python

# Inspect the original data types first
print(orders["order_date"].dtype)

# Display sample values
print(orders["order_date"].head(10))

# Parse dates for inspection only; preserve the raw column
parsed_order_dates = pd.to_datetime(
    orders["order_date"],
    errors="coerce"
)

print("Earliest order date:", parsed_order_dates.min())
print("Latest order date:", parsed_order_dates.max())
print("Unparseable order dates:", parsed_order_dates.isna().sum())
```

Since nonoe of the dates use an ambiguous format, such as day/month/year versus month/day/year, so we don't need to investigate anything further. None of the dates wereunparseable

We don't want to assume that the table (`orders_2024`) contains every transaction through December 31, 2024.
