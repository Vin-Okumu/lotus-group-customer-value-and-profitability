#### Check 6: Identifying Potential Foreign Key Problems

We want to check whether transaction records reference customers, products, stores and employees that actually exist.

**The relationships we expect to exist:**

|Child table	|Foreign key	|Parent table	|Primary key|
|---|---|---|---|
|Orders	|`customer_id`	|Customers	|`customer_id`|
|Orders	|`store_id`	|Stores	|`store_id`|
|Orders	|`employee_id`	|Employees	|`employee_id`|
|Orders	|`date_id`	|Date	|`date_id`|
|Employees	|`store_id`	|Stores	|`store_id`|
|Order details	|`order_id`	|Orders	|`order_id`|
|Order details	|`product_id`	|Products	|`product_id`|
|Returns	|`order_id`	|Orders	|`order_id`|

Notice that `employee_id` is allowed to be missing based on our earlier findings. We will distinguish missing references from invalid references.

```python
relationships = [
    {
        "child": orders,
        "child_column": "customer_id",
        "parent": customers,
        "parent_column": "customer_id",
        "relationship": "Orders -> Customers"
    },
    {
        "child": orders,
        "child_column": "store_id",
        "parent": stores,
        "parent_column": "store_id",
        "relationship": "Orders -> Stores"
    },
    {
        "child": orders,
        "child_column": "employee_id",
        "parent": employees,
        "parent_column": "employee_id",
        "relationship": "Orders -> Employees"
    },
    {
        "child": orders,
        "child_column": "date_id",
        "parent": date,
        "parent_column": "date_id",
        "relationship": "Orders -> Date"
    },
    {
        "child": employees,
        "child_column": "store_id",
        "parent": stores,
        "parent_column": "store_id",
        "relationship": "Employees -> Stores"
    },
    {
        "child": order_details,
        "child_column": "order_id",
        "parent": orders,
        "parent_column": "order_id",
        "relationship": "Order Details -> Orders"
    },
    {
        "child": order_details,
        "child_column": "product_id",
        "parent": products,
        "parent_column": "product_id",
        "relationship": "Order Details -> Products"
    },
    {
        "child": returns,
        "child_column": "order_id",
        "parent": orders,
        "parent_column": "order_id",
        "relationship": "Returns -> Orders"
    }
]

for rel in relationships:

    child = rel["child"]
    parent = rel["parent"]

    child_col = rel["child_column"]
    parent_col = rel["parent_column"]

    valid_keys = set(parent[parent_col].dropna())

    missing_count = child[child_col].isna().sum()

    unmatched = child.loc[
        child[child_col].notna()
        & ~child[child_col].isin(valid_keys),
        child_col
    ]

    print(f"\n{rel['relationship']}")
    print("Total records:", len(child))
    print("Missing foreign keys:", missing_count)
    print("Unmatched records:", len(unmatched))
    print("Distinct unmatched values:", unmatched.nunique())
```

### Investigating Transaction Dates

Before cleaning or transforming the data, we want to establish whether the transaction dates are logically consistent.

In particular, we want to check whether:

1. `order_date` agrees with the corresponding `full_date` in the date dimension through `date_id`.

2. The dates fall within the expected coverage period.

3. Return dates occur on or after their corresponding order dates.

4. Any dates are missing or cannot be interpreted.

Let's start with the most important check: does the order date agree with the date dimension?

```python
# Convert date columns temporarily for validation
orders["order_date_check"] = pd.to_datetime(
    orders["order_date"], errors="coerce"
)

date["full_date_check"] = pd.to_datetime(
    date["full_date"], errors="coerce"
)

# Join orders to the date dimension
date_validation = orders.merge(
    date[["date_id", "full_date_check"]],
    on="date_id",
    how="left"
)

# Compare the two dates
date_validation["date_matches"] = (
    date_validation["order_date_check"].dt.date
    == date_validation["full_date_check"].dt.date
)

# Summarize results
print("Total orders:", len(date_validation))

print(
    "Unparseable order dates:",
    date_validation["order_date_check"].isna().sum()
)

print(
    "Unparseable dimension dates:",
    date_validation["full_date_check"].isna().sum()
)

print(
    "Matching dates:",
    date_validation["date_matches"].sum()
)

print(
    "Mismatching dates:",
    (~date_validation["date_matches"]).sum()
)
```

### Validating Return Dates
Here we want to establish whether returns occur on or after their corresponding order dates. A return dated before its original purchase would be a chronological inconsistency.

```python
# Temporarily convert return dates
returns["return_date_check"] = pd.to_datetime(
    returns["return_date"], errors="coerce"
)

# Attach the original order date to each return
return_validation = returns.merge(
    orders[["order_id", "order_date_check"]],
    on="order_id",
    how="left"
)

# Compare return date against order date
return_validation["return_date_valid"] = (
    return_validation["return_date_check"]
    >= return_validation["order_date_check"]
)

# Summarize results
print("Total return records:", len(return_validation))

print(
    "Unparseable return dates:",
    return_validation["return_date_check"].isna().sum()
)

print(
    "Returns occurring before order date:",
    (
        return_validation["return_date_check"]
        < return_validation["order_date_check"]
    ).sum()
)

print(
    "Valid chronological returns:",
    return_validation["return_date_valid"].sum()
)
```

## Validating Transaction Calculation

This is essential because our revenue, profit and margin figures depend on these calculations being correct.

We want to verify three mathematical relationships.

**Relationship A: Selling price**

- The expected calculation is:

    - Selling price = Unit price × (1 − Discount percentage / 100)

So, suppose:
- Unit price = 1,000 EGP
- Discount = 10%
The expected selling price is 900 EGP.

**Relationship B: Line revenue**

- Line revenue = Quantity × Selling price

**Relationship C: Line cost**

- Line cost = Quantity × Unit cost

Let's test all three.

```python
# Calculate expected values
order_details["expected_selling_price"] = (
    order_details["unit_price"]
    * (1 - order_details["discount_pct"] / 100)
)

order_details["expected_line_revenue"] = (
    order_details["quantity"]
    * order_details["selling_price"]
)

order_details["expected_line_cost"] = (
    order_details["quantity"]
    * order_details["unit_cost"]
)

# Calculate differences
order_details["selling_price_difference"] = (
    order_details["selling_price"]
    - order_details["expected_selling_price"]
)

order_details["revenue_difference"] = (
    order_details["line_total_revenue"]
    - order_details["expected_line_revenue"]
)

order_details["cost_difference"] = (
    order_details["line_total_cost"]
    - order_details["expected_line_cost"]
)

# Check consistency using a small floating-point tolerance
tolerance = 0.01

print("SELLING PRICE VALIDATION")
print(
    "Mismatches:",
    (
        order_details["selling_price_difference"].abs()
        > tolerance
    ).sum()
)

print("\nLINE REVENUE VALIDATION")
print(
    "Mismatches:",
    (
        order_details["revenue_difference"].abs()
        > tolerance
    ).sum()
)

print("\nLINE COST VALIDATION")
print(
    "Mismatches:",
    (
        order_details["cost_difference"].abs()
        > tolerance
    ).sum()
)
```

## Reconciliation of Order Details Against Order-level Financial Records

Here we want to establish whether the sum of individual order lines agrees with the corresponding order's recorded revenue and cost.

This is important because our dataset contains:

- 25,099 order-detail records.

- 12,000 order records.

Each order may contain multiple products. Therefore, summing its individual line revenues and costs should generally reconcile with the financial totals recorded in the corresponding order.

```python
# Aggregate order-detail financial values by order_id
order_detail_totals = (
    order_details
    .groupby("order_id", as_index=False)
    .agg(
        calculated_revenue=("line_total_revenue", "sum"),
        calculated_cost=("line_total_cost", "sum")
    )
)

# Merge aggregated values with order-level records
order_reconciliation = orders.merge(
    order_detail_totals,
    on="order_id",
    how="left"
)

# Calculate differences
order_reconciliation["revenue_difference"] = (
    order_reconciliation["total_revenue"]
    - order_reconciliation["calculated_revenue"]
)

order_reconciliation["cost_difference"] = (
    order_reconciliation["total_cost"]
    - order_reconciliation["calculated_cost"]
)

# Count mismatches
tolerance = 0.01

print("ORDER-LEVEL REVENUE RECONCILIATION")
print(
    "Mismatches:",
    (
        order_reconciliation["revenue_difference"].abs()
        > tolerance
    ).sum()
)

print("\nORDER-LEVEL COST RECONCILIATION")
print(
    "Mismatches:",
    (
        order_reconciliation["cost_difference"].abs()
        > tolerance
    ).sum()
)

print("\nOrders without matching order details:")
print(
    order_reconciliation["calculated_revenue"].isna().sum()
)
```

## Order Status Investigation

We want to establish what constitutes a qualifying purchase.

Our classification target is:

- 1: Customer makes at least one qualifying purchase within the next 90 days.

- 0: Customer makes no qualifying purchase within the next 90 days.

We also want to establish which transactions should contribute to the regression target: total customer spending over the same 90-day period.

Our dataset contains three order statuses:

- Completed

- Returned

- Cancelled

We don't want to automatically assume that all three represent actual purchases.

```python
# Examine order statuses and their financial values
order_status_summary = (
    orders
    .groupby("order_status")
    .agg(
        order_count=("order_id", "count"),
        unique_customers=("customer_id", "nunique"),
        total_revenue=("total_revenue", "sum"),
        average_order_value=("total_revenue", "mean"),
        total_cost=("total_cost", "sum")
    )
    .reset_index()
)

print(order_status_summary)
```

```python
# Examining return statuses
return_status_summary = (
    returns
    .groupby("return_status")
    .agg(
        return_count=("return_id", "count"),
        total_return_amount=("return_amount", "sum")
    )
    .reset_index()
)

print(return_status_summary)
```

## Checking Whether Rreturn Records Correspond to Returned Orders

We want to establish whether orders with a Returned status actually have corresponding return records, and whether refunded amounts are consistent with the order values.

```python
# Summarise return records by order
return_order_summary = (
    returns
    .groupby("order_id")
    .agg(
        return_records=("return_id", "count"),
        total_refund=("return_amount", "sum"),
        refunded_amount=(
            "return_amount",
            lambda x: x[returns.loc[x.index, "return_status"] == "Refunded"].sum()
        )
    )
    .reset_index()
)

# Join return information to orders
return_check = orders.merge(
    return_order_summary,
    on="order_id",
    how="left"
)

# Replace missing return counts with zero
return_check["return_records"] = (
    return_check["return_records"].fillna(0)
)

print("ORDERS WITH RETURN STATUS")
print(
    "Returned orders:",
    (return_check["order_status"] == "Returned").sum()
)

print(
    "Returned orders without return records:",
    (
        (return_check["order_status"] == "Returned")
        & (return_check["return_records"] == 0)
    ).sum()
)

print(
    "Completed orders with return records:",
    (
        (return_check["order_status"] == "Completed")
        & (return_check["return_records"] > 0)
    ).sum()
)
```

## Customer Dates Validation

We need reliable birth dates and registration dates because they may help us derive:

- Customer age.

- Customer tenure (how long they have been registered).

- Customer lifecycle characteristics.

```python
# Convert date columns temporarily for validation
customers["birth_date_check"] = pd.to_datetime(
    customers["birth_date"],
    errors="coerce"
)

customers["registration_date_check"] = pd.to_datetime(
    customers["registration_date"],
    errors="coerce"
)

print("CUSTOMER DATE VALIDATION")

print(
    "Invalid birth dates:",
    customers["birth_date_check"].isna().sum()
)

print(
    "Invalid registration dates:",
    customers["registration_date_check"].isna().sum()
)

print(
    "Birth dates after registration:",
    (
        customers["birth_date_check"]
        > customers["registration_date_check"]
    ).sum()
)

print(
    "Registration dates after last available transaction:",
    (
        customers["registration_date_check"]
        > orders["order_date_check"].max()
    ).sum()
)
```
